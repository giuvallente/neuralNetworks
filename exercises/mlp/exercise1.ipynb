### Exercise 1 - One Step of Backpropagation, by Hand

A — Forward pass

We apply $\tanh$ to both the hidden layer and the output neuron. Calculations retain full precision; displayed results are rounded to 6 decimal places.

**1. Hidden pre-activations**

$$
\mathbf{z}^{(1)}=\mathbf{W}^{(1)}\mathbf{x}+\mathbf{b}^{(1)}
$$

For the first hidden neuron:

$$
z_1^{(1)}
=(0.3000)(0.5000)+(-0.1000)(-0.2000)+0.1000
=0.1500+0.0200+0.1000
=\boxed{0.270000}
$$

For the second hidden neuron:

$$
z_2^{(1)}
=(0.2000)(0.5000)+(0.4000)(-0.2000)-0.2000
=0.1000-0.0800-0.2000
=\boxed{-0.180000}
$$

Therefore,

$$
\mathbf{z}^{(1)}=
\begin{bmatrix}
0.270000\\
-0.180000
\end{bmatrix}
$$

**2. Hidden activations**

Using $\tanh(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}$:

$$
h_1^{(1)}=\tanh(0.270000)\approx\boxed{0.263625}
$$

$$
h_2^{(1)}=\tanh(-0.180000)\approx\boxed{-0.178081}
$$

Thus,

$$
\mathbf{h}^{(1)}\approx
\begin{bmatrix}
0.263625\\
-0.178081
\end{bmatrix}
$$

**3. Output pre-activation**

$$
u^{(2)}=\mathbf{W}^{(2)}\mathbf{h}^{(1)}+b^{(2)}
$$

$$
u^{(2)}
=(0.5000)\tanh(0.270000)
+(-0.3000)\tanh(-0.180000)
+0.2000
$$

$$
u^{(2)}
\approx0.131812+0.053424+0.200000
\approx\boxed{0.385237}
$$

**4. Prediction and loss**

The output activation gives:

$$
\hat{y}=\tanh(u^{(2)})
=\tanh(0.3852366782)
\approx\boxed{0.367247}
$$

Since $N=1$ and $y=1.0000$, the mean squared error is:

$$
L=(y-\hat{y})^2
=(1.0000-0.3672465626)^2
$$

$$
L\approx(0.6327534374)^2
=\boxed{0.400377}
$$

The network’s **prediction is $0.367247$**, and its **loss is $0.400377$**.

B — Backward pass

Using the values from item A, we apply the chain rule to compute every gradient. Calculations retain full precision; displayed results are rounded to 6 decimal places.

**1. Loss gradient and output pre-activation gradient**

Since $L=(y-\hat y)^2$,

$$
\frac{\partial L}{\partial \hat y}
=2(\hat y-y)
=2(0.3672465626-1.0000)
=\boxed{-1.265507}
$$

The derivative of the output activation is:

$$
\frac{\partial \hat y}{\partial u^{(2)}}
=1-\tanh^2(u^{(2)})
=1-\hat y^2
\approx\boxed{0.865130}
$$

Therefore,

$$
\frac{\partial L}{\partial u^{(2)}}
=
\frac{\partial L}{\partial\hat y}
\frac{\partial\hat y}{\partial u^{(2)}}
$$

$$
\approx(-1.265507)(0.865130)
=\boxed{-1.094828}
$$

Let $\delta^{(2)}=\frac{\partial L}{\partial u^{(2)}}$.

**2. Output-layer gradients**

Because $u^{(2)}=W_1^{(2)}h_1^{(1)}+W_2^{(2)}h_2^{(1)}+b^{(2)}$,

$$
\frac{\partial L}{\partial W_j^{(2)}}
=\delta^{(2)}h_j^{(1)}
$$

Thus,

$$
\frac{\partial L}{\partial W_1^{(2)}}
\approx(-1.094828)(0.263625)
=\boxed{-0.288624}
$$

$$
\frac{\partial L}{\partial W_2^{(2)}}
\approx(-1.094828)(-0.178081)
=\boxed{0.194968}
$$

$$
\boxed{
\frac{\partial L}{\partial\mathbf W^{(2)}}
\approx
\begin{bmatrix}
-0.288624 & 0.194968
\end{bmatrix}
}
$$

Since $\frac{\partial u^{(2)}}{\partial b^{(2)}}=1.0000$,

$$
\boxed{\frac{\partial L}{\partial b^{(2)}}\approx-1.094828}
$$

**3. Propagate back through the hidden activations**

For each hidden neuron,

$$
\frac{\partial L}{\partial h_j^{(1)}}
=\delta^{(2)}W_j^{(2)}
$$

$$
\frac{\partial L}{\partial h_1^{(1)}}
\approx(-1.094828)(0.5000)
=\boxed{-0.547414}
$$

$$
\frac{\partial L}{\partial h_2^{(1)}}
\approx(-1.094828)(-0.3000)
=\boxed{0.328448}
$$

Next, apply the hidden-layer $\tanh$ derivative:

$$
\frac{\partial h_1^{(1)}}{\partial z_1^{(1)}}
=1-\tanh^2(0.270000)
\approx\boxed{0.930502}
$$

$$
\frac{\partial h_2^{(1)}}{\partial z_2^{(1)}}
=1-\tanh^2(-0.180000)
\approx\boxed{0.968287}
$$

Therefore,

$$
\frac{\partial L}{\partial z_1^{(1)}}
\approx(-0.547414)(0.930502)
=\boxed{-0.509370}
$$

$$
\frac{\partial L}{\partial z_2^{(1)}}
\approx(0.328448)(0.968287)
=\boxed{0.318032}
$$

Let

$$
\boldsymbol\delta^{(1)}
=\frac{\partial L}{\partial\mathbf z^{(1)}}
\approx
\begin{bmatrix}
-0.509370\\
0.318032
\end{bmatrix}.
$$

**4. Hidden-layer gradients**

Since $z_i^{(1)}=\sum_j W_{ij}^{(1)}x_j+b_i^{(1)}$,

$$
\frac{\partial L}{\partial W_{ij}^{(1)}}
=\delta_i^{(1)}x_j
$$

The four weight gradients are:

$$
\frac{\partial L}{\partial W_{11}^{(1)}}
\approx(-0.509370)(0.5000)
=\boxed{-0.254685}
$$

$$
\frac{\partial L}{\partial W_{12}^{(1)}}
\approx(-0.509370)(-0.2000)
=\boxed{0.101874}
$$

$$
\frac{\partial L}{\partial W_{21}^{(1)}}
\approx(0.318032)(0.5000)
=\boxed{0.159016}
$$

$$
\frac{\partial L}{\partial W_{22}^{(1)}}
\approx(0.318032)(-0.2000)
=\boxed{-0.063606}
$$

Thus,

$$
\boxed{
\frac{\partial L}{\partial\mathbf W^{(1)}}
\approx
\begin{bmatrix}
-0.254685 & 0.101874\\
0.159016 & -0.063606
\end{bmatrix}
}
$$

The bias gradients equal the corresponding pre-activation gradients:

$$
\boxed{
\frac{\partial L}{\partial\mathbf b^{(1)}}
\approx
\begin{bmatrix}
-0.509370\\
0.318032
\end{bmatrix}
}
$$

C — Parameter update

With learning rate $\eta=0.3000$, each parameter is updated using

$$
\theta_{\text{new}}
=\theta_{\text{old}}-0.3000\frac{\partial L}{\partial\theta}.
$$

All updates use the gradients evaluated at the original parameters.

**1. Output-layer updates**

$$
W_{1,\text{new}}^{(2)}
\approx0.5000-0.3000(-0.288624)
=\boxed{0.586587}
$$

$$
W_{2,\text{new}}^{(2)}
\approx-0.3000-0.3000(0.194968)
=\boxed{-0.358490}
$$

$$
b_{\text{new}}^{(2)}
\approx0.2000-0.3000(-1.094828)
=\boxed{0.528448}
$$

Therefore,

$$
\boxed{
\mathbf W_{\text{new}}^{(2)}
\approx
\begin{bmatrix}
0.586587 & -0.358490
\end{bmatrix},
\qquad
b_{\text{new}}^{(2)}\approx0.528448
}
$$

**2. Hidden-layer updates**

$$
W_{11,\text{new}}^{(1)}
\approx0.3000-0.3000(-0.254685)
=\boxed{0.376405}
$$

$$
W_{12,\text{new}}^{(1)}
\approx-0.1000-0.3000(0.101874)
=\boxed{-0.130562}
$$

$$
W_{21,\text{new}}^{(1)}
\approx0.2000-0.3000(0.159016)
=\boxed{0.152295}
$$

$$
W_{22,\text{new}}^{(1)}
\approx0.4000-0.3000(-0.063606)
=\boxed{0.419082}
$$

$$
b_{1,\text{new}}^{(1)}
\approx0.1000-0.3000(-0.509370)
=\boxed{0.252811}
$$

$$
b_{2,\text{new}}^{(1)}
\approx-0.2000-0.3000(0.318032)
=\boxed{-0.295410}
$$

Therefore,

$$
\boxed{
\mathbf W_{\text{new}}^{(1)}
\approx
\begin{bmatrix}
0.376405 & -0.130562\\
0.152295 & 0.419082
\end{bmatrix},
\qquad
\mathbf b_{\text{new}}^{(1)}
\approx
\begin{bmatrix}
0.252811\\
-0.295410
\end{bmatrix}
}
$$

**3. Effect on the loss**

A second forward pass with the updated parameters gives:

$$
\mathbf z_{\text{new}}^{(1)}
=\mathbf W_{\text{new}}^{(1)}\mathbf x+\mathbf b_{\text{new}}^{(1)}
\approx
\begin{bmatrix}
0.467126\\
-0.303079
\end{bmatrix}
$$

$$
\mathbf h_{\text{new}}^{(1)}
=\tanh(\mathbf z_{\text{new}}^{(1)})
\approx
\begin{bmatrix}
0.435874\\
-0.294127
\end{bmatrix}
$$

$$
u_{\text{new}}^{(2)}
=\mathbf W_{\text{new}}^{(2)}\mathbf h_{\text{new}}^{(1)}
+b_{\text{new}}^{(2)}
\approx0.889568
$$

$$
\hat y_{\text{new}}
=\tanh(u_{\text{new}}^{(2)})
\approx\boxed{0.711181}
$$

$$
L_{\text{new}}
=(1.0000-0.7111805295)^2
\approx\boxed{0.083417}
$$

The loss decreases from $0.400377$ to $0.083417$ because the update moves the prediction closer to the target $y=1.0000$. Gradient descent follows the direction of steepest local decrease; the second forward pass confirms that this learning rate produces a decrease here.
